# Analiza Danych Szkolnych
## Wczytywanie i analiza plików: Student.csv, Teacher.csv, Section.csv, StudentEnrollment.csv, TeacherRoster.csv

```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns

# Konfiguracja
INPUT = Path("/mnt/user-data/uploads")
OUTPUT = Path("/mnt/user-data/outputs")

# Styl wykresów
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)
```

## 1. Wczytanie danych

```python
# Wczytaj dane studentów
students_df = pd.read_csv(INPUT / "Student.csv")
print(f"Studenci: {len(students_df)} wierszy")
display(students_df.head())
print("\nInformacje o kolumnach:")
display(students_df.info())
```

```python
# Wczytaj dane nauczycieli
teachers_df = pd.read_csv(INPUT / "Teacher.csv")
print(f"Nauczyciele: {len(teachers_df)} wierszy")
display(teachers_df.head())
print("\nInformacje o kolumnach:")
display(teachers_df.info())
```

```python
# Wczytaj dane sekcji
sections_df = pd.read_csv(INPUT / "Section.csv")
print(f"Sekcje: {len(sections_df)} wierszy")
display(sections_df.head())
print("\nInformacje o kolumnach:")
display(sections_df.info())
```

```python
# Wczytaj zapisy studentów
enrollments_df = pd.read_csv(INPUT / "StudentEnrollment.csv")
print(f"Zapisy studentów: {len(enrollments_df)} wierszy")
display(enrollments_df.head())
print("\nInformacje o kolumnach:")
display(enrollments_df.info())
```

```python
# Wczytaj rozkład nauczycieli
rosters_df = pd.read_csv(INPUT / "TeacherRoster.csv")
print(f"Rozkład nauczycieli: {len(rosters_df)} wierszy")
display(rosters_df.head())
print("\nInformacje o kolumnach:")
display(rosters_df.info())
```

## 2. Podstawowe statystyki

```python
print("PODSUMOWANIE DANYCH:")
print("=" * 60)
print(f"Liczba studentów: {len(students_df)}")
print(f"Liczba nauczycieli: {len(teachers_df)}")
print(f"Liczba sekcji: {len(sections_df)}")
print(f"Liczba zapisów: {len(enrollments_df)}")
print(f"Liczba przypisań nauczycieli: {len(rosters_df)}")
```

## 3. Analiza relacji między tabelami

```python
# Przykład: Liczba studentów w każdej sekcji
if 'SectionID' in enrollments_df.columns:
    students_per_section = enrollments_df.groupby('SectionID').size().reset_index(name='StudentCount')
    print("Liczba studentów w każdej sekcji:")
    display(students_per_section.describe())
    
    # Wizualizacja
    plt.figure(figsize=(12, 6))
    plt.hist(students_per_section['StudentCount'], bins=20, edgecolor='black')
    plt.xlabel('Liczba studentów')
    plt.ylabel('Liczba sekcji')
    plt.title('Rozkład liczby studentów w sekcjach')
    plt.show()
```

## 4. Plik PDF: przypisania.pdf

```python
# Sprawdź plik PDF
pdf_file = INPUT / "przypisania.pdf"
if pdf_file.exists():
    print(f"✓ Plik przypisania.pdf istnieje")
    print(f"  Rozmiar: {pdf_file.stat().st_size} bajtów")
    print(f"  Ścieżka: {pdf_file}")
else:
    print("⚠ Plik przypisania.pdf nie został znaleziony")
```

## 5. Eksport wyników

```python
# Zapisz przetwortzone dane do katalogu wyjściowego
OUTPUT.mkdir(parents=True, exist_ok=True)

# Przykład: zapisz podsumowanie
summary = {
    'Dataset': ['Students', 'Teachers', 'Sections', 'Enrollments', 'Rosters'],
    'RowCount': [len(students_df), len(teachers_df), len(sections_df), len(enrollments_df), len(rosters_df)],
    'ColumnCount': [len(students_df.columns), len(teachers_df.columns), len(sections_df.columns), 
                    len(enrollments_df.columns), len(rosters_df.columns)]
}

summary_df = pd.DataFrame(summary)
summary_df.to_csv(OUTPUT / "data_summary.csv", index=False)
print("Podsumowanie zapisano do:", OUTPUT / "data_summary.csv")
display(summary_df)
```
